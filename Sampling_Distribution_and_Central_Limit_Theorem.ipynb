```python
# Sampling Distribution of Mean --> Describes how a statistic varies across repeated samples
# of the sample size from population.

# [40,50,60,70,80] --> [40,50] --> 45
# [40,60] --> 50, [40,70] --> 55, [40,80] --> 60, [50,60] --> 55

# [50,70] --> 60, [50,80] --> 65, [60,70] --> 65, [60,80] --> 70, [70,80] --> 75

# --> mean
#Each is equally likely under simple random sampling
#Avg of all 10 sample means = 45+50+55+60+55+60+65+70+75/10. --> 600/10 = 60

#The sample mean is an unbiased estimator of the Population mean..
#This doesn't tell every sample mean is equal to population mean...

#Population distribution --> individual values in Population
#sample distribution --> individual values in sample
#Sampling distribution of mean --> means from resp samples

from itertools import combinations
import statistics
population = [40,50,60,70,80]
means = []
print(list(combinations(population,2)))
print(len(list(combinations(population,2))))
```

```python
for sample in combinations(population,2):
  means.append(statistics.mean(sample))
print(means)
print("Mean of Means:",statistics.mean(means))
print("Population of Mean:",statistics.mean(population))
```

```python
#Central limit theorem --> for independent, identically distributed observations with finite variance --> CLT it states says that sampling dstn of the mean becomes apprx
#normal as sample size increases

#Centre --> mu
#Variance --> sigma(squared) / n
#Std deviation of sample means = sigma / sqrt.n

#n = 1 -->Sample means will follows the poulation dstn
#n = 5 --> it may still noticeably skewed
#n = 30 --> it is often more apprx normal
#n = 100 --> usually closer to normal
```

```python
population = [40,50,60,70,80]
means = []
print(list(combinations(population,3)))
for sample in combinations(population,3):
  means.append(statistics.mean(sample))
print(means)
print("Mean of Means:",statistics.mean(means))
print("Population of Mean:",statistics.mean(population))
```

```python
#Standard Deviation and Standard Error (Variation of an estimate)--> How much sample average vary...
import math
import statistics
scores = [55,60,65,70,75]
sample_size = len(scores)
sample_mean = statistics.mean(scores)
print(sample_mean) #Estimate of population mean
sample_std = statistics.stdev(scores)
print(sample_std)
estimated_se = sample_std / math.sqrt(sample_size)  #We are getting to the possible sample
print(estimated_se)
print(f'Sample Size{sample_size}')
print(f'Sample Mean{sample_mean}')
print('Sample SD:',round(sample_std,3))
print("Estimated SE",round(estimated_se,3))
```

```python
#Sample sizes --> Standard Error
population_std = 12
sample_sizes = [9,36,44]
for sample_size in sample_sizes:
  se = population_std / math.sqrt(sample_size)
  print(se)
  print("n = ",sample_size,"SE = ",round(se,3))
```
